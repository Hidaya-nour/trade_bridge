## 4. Outlier Detection and Treatment

Detect outliers using statistical methods like IQR or visualization, and treat them by capping, removing, or transforming.

```python
# Outlier detection using IQR
def detect_outliers_iqr(data):
    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return (data < lower_bound) | (data > upper_bound)

# Example for a numeric column
# outliers = detect_outliers_iqr(df['demand'])
# print(f"Number of outliers in demand: {outliers.sum()}")

# Treat outliers (e.g., cap them)
# df['demand'] = np.where(df['demand'] > upper_bound, upper_bound, df['demand'])
```

## 7. Feature Selection

Select relevant features using techniques like correlation analysis or recursive feature elimination from scikit-learn.

```python
from sklearn.feature_selection import SelectKBest, f_regression

# Assuming 'demand' is the target
X = df.drop('demand', axis=1).select_dtypes(include=['float64', 'int64'])
y = df['demand']

# Select top k features
selector = SelectKBest(score_func=f_regression, k=10)
X_selected = selector.fit_transform(X, y)

print("Selected features shape:", X_selected.shape)
print("Feature scores:", selector.scores_)
```

## 8. Split the Dataset

Split the dataset into training and testing sets using train_test_split from scikit-learn for model evaluation.

```python
from sklearn.model_selection import train_test_split

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(X_selected, y, test_size=0.2, random_state=42)

print("Training set shape:", X_train.shape)
print("Testing set shape:", X_test.shape)
```
